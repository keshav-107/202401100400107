```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
X = np.array([1, 2, 3, 4, 5], dtype=float).reshape(-1, 1)
y = np.array([2, 4, 6, 8, 10], dtype=float).reshape(-1, 1)
y_norm=[]
for i in y:
    y_norm.append(i/10)
print(y_norm)

lr = 0.001
epochs = 10000
w = np.random.randn(1, 1)
b = np.random.randn(1, 1)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def sigmoid_derivative(z):
    return sigmoid(z) * (1 - sigmoid(z))
```

```python
loss_history = []

best_loss = float('inf')
best_w, best_b = None, None

for epoch in range(epochs):

    y_pred = X @ w + b

    loss = np.mean((y - y_pred) ** 2)
    loss_history.append(loss)

    dw = (2 / len(X)) * (X.T @ (y_pred - y))
    db = (2 / len(X)) * np.sum(y_pred - y)

    w -= lr * dw
    b -= lr * db

    if loss < best_loss:
        best_loss = loss
        best_w = w.copy()
        best_b = b.copy()

    if (epoch + 1) % 500 == 0:
        print(f"Epoch {epoch+1}/{epochs} | Loss: {loss:.6f}")
```

```python
test_x = 10
z_test = test_x * w + b
pred_norm = sigmoid(z_test)
pred_actual = pred_norm * 10
```

```python
print(f"--- Training Complete ---")
# print(f"Final Loss: {loss_history}")
print("Best Weight:", best_w)
print("Best Bias:", best_b)
print("Minimum Loss:", best_loss)
print(f"Predicted y for x={test_x}: {pred_actual}")
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
z_range = np.linspace(-10, 10, 100)
plt.plot(z_range, sigmoid(z_range), color='blue', lw=2)
plt.title("Sigmoid Activation Function")
plt.grid(True, linestyle='--', alpha=0.6)
```

```python
plt.plot(loss_history)
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Loss vs Epoch")
plt.show()
```
